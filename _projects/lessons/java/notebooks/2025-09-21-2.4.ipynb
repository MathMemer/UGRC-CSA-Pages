# 2.4 Nested `if` / `else if` Statements

**Learning objective:** Trace and write `if / else if / else` decision chains to determine which single branch executes for a given input.

**Success criteria — I can:**
- Trace which branch of an `else if` chain runs for a given input.
- Explain why only one branch in the chain executes, even when more than one condition would be true on its own.
- Write my own `else if` chain that classifies a value into categories.

This is the same branching pattern you'll use to route input or apply business rules in your own projects — just with more than two outcomes.

## Engage (first 5 minutes) — predict first, don't click Run yet!

With `myAge = 17`, which lines below will print? Tell a partner your prediction, then click **Run** on the example below to check.

```java
// CODE_RUNNER: Predict the output for myAge = 17, then change myAge and rerun.
public class Main {
    public static void main(String[] args) {
        int myAge = 17; // adjust the age to see the difference in what it prints out
        System.out.println("Current age: " + myAge);
        if (myAge >= 18)
        {
            System.out.println("You can register to vote.");
            System.out.println("You are old enough for a license to drive.");
        }
        else if (myAge >= 16) {
            System.out.println("You are old enough for a license to drive.");
        }
        else if (myAge >= 15) {
            System.out.println("You can learn to drive a car.");
        }
        else if (myAge >= 14) {
            System.out.println("You are 14");
        }
    }
}

Main.main(null);
```

## Tech Talk

**Else if statements** are used when you have multiple conditions that need to be checked sequentially.

**Flow of execution:** each condition is evaluated in the order written. The first true condition's code runs, and the rest are skipped.

**Structure:**
- Start with a single `if` statement.
- Follow with as many `else if` statements as needed.
- Optionally end with one `else` to handle any remaining cases.

**Key concept:** order matters — a broader condition placed earlier in the chain will "shadow" a more specific condition later on, so it will never run.

## Popcorn Hack (in class)

The example below assigns a letter grade using the same `else if` pattern. Add more `else if` branches so the code also prints out an **"F"** and a **"C"**.

```java
// CODE_RUNNER: Add else if branches so the code also prints "C" and "F".
public class gradeEvaluator
{
    public static void main(String[] args)
        {
        double myAverage = 90;
        System.out.println("Current average: " + myAverage);
        if (myAverage >= 90)
        {
            System.out.println("You have \"A\" average.");
        }

        else if (myAverage >= 80)
        {
            System.out.println("You have an \"B\" average.");
        }
        //add more letter grade options as needed
    }
}

gradeEvaluator.main(null);
```
